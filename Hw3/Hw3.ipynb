```python
import os
from ultralytics.hub.utils import HUB_API_ROOT
from ultralytics.utils.downloads import download
from pathlib import Path

# Удаляем прошлые неудачные попытки скачивания
!rm -rf /content/datasets

# Скачиваем архив напрямую через системную функцию утилит Ultralytics
url = 'https://github.com/ultralytics/assets/releases/download/v0.0.0/homeobjects-3K.zip'
download(url, dir=Path('/content/datasets'))

# Проверяем, что папки теперь действительно существуют
print("Проверка путей:")
print("Train images:", os.path.exists('/content/datasets/homeobjects-3K/images/train'))
print("Val images:", os.path.exists('/content/datasets/homeobjects-3K/images/val'))
```

```python
from ultralytics import YOLO

# Загружаем предобученную модель
model = YOLO('yolov5s.pt')

# Запускаем обучение на исправленном датасете
results = model.train(
    data='HomeObjects-3K.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    device=0
)
```

```python
from ultralytics import YOLO

# 1. Загружаем обученную модель
model = YOLO('runs/detect/train/weights/best.pt')

# 2. Передаем имя папки 'img' для пакетной обработки всех картинок внутри
results = model.predict(
    source='img',     # Путь к вашей папке из файлового менеджера
    conf=0.25,        # Порог уверенности детекции
    save=True         # Сохранить размеченные картинки на диск
)
```
